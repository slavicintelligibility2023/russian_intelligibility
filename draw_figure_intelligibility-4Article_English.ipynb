## INTRODUCTION
It has long been recognized that speakers of related languages can have successful communication speaking their native languages. Various terms have been used to refer to such kind of communication like semicommunication, receptive multilingualism, semibilingualism, inherent intelligibility, non-convergent discourse, asymmetric/bilingual discourse (Gooskens, 2013).

Mutual intelligibility has long been used as one of the criteria to identify whether two speech varieties should be considered dialects of the same language or two distinct languages (Gleason 1959, Comrie 1981). Knowledge about the degree to which two related languages are mutually intelligible is also required for various language planning, language policy, language contact issues.

Receptive multilingualism has recently come into focus as a way of communicating across language borders, a possible alternative to using a lingua franca. Two major projects dedicated to mutual intelligibility studies need to be mentioned here. 

The first one, MICReLa (Mutual Intelligibility of Closely Related Languages) described in Gooskens & Swarte (2017), Gooskens et al. (2018), focuses on receptive multilingualism in Europe aiming to improve communication in the region. The project studies what factors influence mutual intercomprehension and to what extent it can be used to overcome language barriers. It follows the team’s preceding research into intelligibility between Scandinavian languages and currently covers 16 languages spoken in Europe from Germanic, Romance, and Slavic language families. Though this large-scale study provides a unified approach and an important basis for comparison of intelligibility within language families, the resulting picture of mutual intelligibility between languages is incomplete. The Slavic language family in this project is represented by the West and South Slavic languages. As we demonstrate below, the missing information on the East Slavic languages (and most importantly Russian) can help to understand why Bulgarian is the least intelligible to other Slavic languages in this study.

The second major project, INCOMSLAV (Mutual Intelligibility and Surprisal in Slavic Intercomprehension, (Mosbach et al. (2019), Stenger et al. (2020b)), aims to research the relation between information density, encoding density and grammaticalisation in a cross-linguistic perspective with a focus on mutual intelligibility between Slavic languages. The project includes spoken and written intercomprehension experiments between 11 Slavic languages (including Russian) through a web-based platform. At the current stage intelligibility is tested on three levels: words, noun phrases, and sentences. Our research studies the comprehension of words in a text, which supposedly may provide additional semantic clues (other than the ones within the sentence) to facilitate the understanding of missing words in it.

Our paper presents a study of written intelligibility of eight Slavic languages to native speakers of Russian. It aims to check how intelligible words of an unknown but related language can be when given in a text. 

Our first hypothesis was that a parallel text in an unknown Slavic language increases the correctness of answers when people fill gaps in a text in Russian. Presumably this increase should correlate with the intelligibility of the parallel text language.

Our second hypothesis was that the intelligibility of a text in an unknown but related foreign language depends not only on the percentage of words in this text that are historically related with words in one’s native language and still keep their phonetic/graphic and semantic similarity with them, but also on the percentage of other categories of words such as those which look similar but have different meanings.


## Experiment set-up

In our experiment participants were given a text in Russian with omitted words and a parallel text in a Slavic language (unknown to them) to use as a clue. In our case, there were no restrictions on the answers other than the ones imposed by the context in Russian and the clue from a corresponding word in another Slavic language (provided the participant identified it correctly in the sentence). This way of arranging the experiment does not limit participants in their choice of words and allows gathering objective quantitative information on a text intelligibility. 

When choosing words to omit in the test, we selected nouns, verbs, adjectives and adverbs. It was clear that some words would be more difficult to restore than others, so we focused on making the test pieces more or less similar in the respective amount of words easier and more difficult to restore.

If a word was omitted in the text once, it was to be omitted in all other occurrences in this test (so that it would not be possible to guess it in case its equivalent is repeated in the parallel text). Words (totalling 186 lemmas, 232 word occurrences) were omitted in the Russian part of a test independently from translations.

Of the nine languages used in the study, five (including Russian) use Cyrillic alphabet and four use Latin alphabet. To avoid problems with the extended Latin script, we transliterated such parallel texts into Cyrillic using simple rules of substitution, like i-и, p-п, č-ч, sz-ш, szcz-щ etc (For phonetic correspondences we used [Wiktionary](https://en.wiktionary.org/wiki/Appendix:Polish_pronunciation])). For the sake of consistency all parallel texts written in Cyrillic script were transliterated into standard Latin. Thus, participants saw each parallel text written both in Latin and Cyrillic scripts (one above the other) aligned by sentences.

It is possible to restore words in a gapped text using the text redundancy on the basis of limitations imposed by the context itself (Jagrova & Avgustinova (2019), Krutchenko et al. (2020), Lopukhina et al. (2021)). To separate this influence from that of a parallel language text and to create a benchmark for comparison each test piece was given to a control group of Russian native speakers to do without a parallel text in another Slavic language.

All participants did tests on a website without any observation of their behaviour. At the beginning of test each participant was asked to select their level of education, to indicate if they knew any other Slavic language, and if their education included any language studies.1 This information was used to ensure that participants received a parallel text in an unknown language - all Slavic languages named by the participant were excluded from the list from which the parallel language for the test was randomly selected. It was also possible to skip these questions, then the subject was given a control test (one of the six gapped text pieces in Russian without a parallel text).

In the experiment we used three pieces from M. Bulgakov’s Master and Margaret (originally written in Russian), three pieces of H. Sienkiewicz’s Quo Vadis? (originally written in Polish), and their modern translations into the other languages in the study.

In our experiment the correctness of answers was evaluated according to the following scale.
* A word from the original text or its close synonym were considered as a ‘completely correct’ answer. Such answers were assigned 1 point. In the example below such words as замкнуться ‘withdraw into oneself’, отгородиться ‘shut oneself off’ were considered completely correct.
Поэтому Иван и от этого пути отказался, решив избрать третий: _(замкнуться)_ в гордом молчании.
…so Ivan rejected that course too, deciding to adopt the third: he would _(wrap himself)_ in proud silence.
* Words and their paraphrases with the same root but of a different part of speech and other words of the same part of speech which did not make the narrative illogical (i.e. words which did not render the phrase ungrammatical or make it lose sense, but the author was talking about something different there) were considered as ‘partially correct’, e.g., for the given example a noun like замкнутость ‘withdrawal’1 or such verbs as сидеть ‘sit’, замереть ‘freeze’, ждать ‘wait’ were considered ‘partially correct’. In such cases we believed that participants were close in their guess, but did not quite get the sense of the omitted word. Such answers were assigned 0.5 points.
* A blank field, an incorrect part of speech which renders the phrase ungrammatical, and words which make the sentence illogical (i.e. make it lose sense) were considered as ‘incorrect’. For example, одиночество ‘loneliness’ was incorrect in the example above. Such answers received 0 points.

## Description of Data Structure

The following files are placed to the *data* folder. The first number in every file is the row id.

- **user_data.csv** - participants' data. Fields description:
    - *id_user* - identifier of a participant.
    - *test_id* - identifier of a test passed by the participant.
    - *parallel_lang* - a language used for a parallel text. *No Parallel Text* denotes the control group, i.e. those participants who restored words without the parallel text. 
    - *age* - age group of the participant. *sch* denotes school children, bak - bachelor students, mag - master student, fin - finished his or her education.
    - *speciality* - the participant's education included any language studies (translation, linguistics, philology): yes - includes, no - don't includes or was not selected.
    - *known_langs* - if the participant knows any other Slavic language. Five and more languages denote situation when the participant declined any parallel text and volonteered to be in the control group.
    - *ans_cnt* - number of non-empty answers for this participant.
    
    
- **user_answers.csv** - participants' answers. Fields description:
    - *id_user* - identifier of a participant; equals to the same field in the *user_data* table.
    - *test_id* - identifier of a test passed by the participant.
    - *parallel_lang* - a language used for a parallel text; equals to the same field in the *user_data* table.
    - *answer_no* - position of a omitted word in the test.
    - *answer* - participant's answer.
    - *correctness* - asessors' evaluation of the answer. incorr denotes an incorrect answer, pcorr - partially correct anser, corr - correct answer.
    
    
- **texts.csv** - text phragments used as tests. The words, marked by the underscore denote omitted words in parallel texts. All texts are aligned by sentences, i.e. have the same identifier for different languages. Fields description:
    - *id_set* - identifier of a text.
    - *lang_name* - language of the text.
    - *sent_pos* - position of a sentence in the text.
    - *sent* - text of the sentence.

### Проверка достаточности количества респондентов

Попробуем проверить достаточность числа респондентов при помощи z-test.

````python
from statsmodels.stats.weightstats import ztest
from scipy.stats import norm
from scipy.stats import kstest
````

Проверим нормальное ли у нас распределение при помощи теста Колмогорова-Смирнова.
Каждое распределение необходимо привести к единичному: центр в 0, дисперсия равна 1. Только после этого можно проводить тест.

Если значение теста <0.05, то это не нормальное распределение.

Значения среднего и дисперсии считаются ниже.

````python
langs_list11 = ['No Parallel Text', 'Ukranian', 'Belarussian', 'Bulgarian', 'Serbian', 'Polish', 'Czech', 'Slovak', 'Slovene']
````

````python
# Список языков, для которых результат не очень похож на нормальное распределение.
for test in range(1, 7):
    for lang in langs_list11:
        data33 = user_data[['parallel_lang', 'mean_mark']] \
                          [(user_data.test_id==test)& \
                           (user_data.parallel_lang==lang)]['mean_mark']
        if data33.shape[0] == 0:
            continue
        data3 = (data33 - data33.mean()) / data33.std()
        s, p = kstest(data3, 'norm')
        if p < 0.05:
            print(lang, test, f"{p:5.3}")
````

Вроде получается нормально.

При помощи z-test посмотрим, выполняется ли гипотеза о сходстве средних значений распределений.

````python
# Проверяем, не похоже ли полученное распределение на такое же, но без подсказок.
for test in range(1, 7):
    no_par_z = user_data[(user_data.test_id==test)&
                         (user_data.parallel_lang=='No Parallel Text')]['mean_mark']
    for lang in langs_list11[1:]:
        ukr_z = user_data[(user_data.test_id==test)&
                          (user_data.parallel_lang==lang)]['mean_mark']
        if ukr_z.shape[0] == 0:
            continue
        s, p = ztest(no_par_z, ukr_z)
        if p > 0.05:
            print(lang, test, f"{p:5.3}")
````

Теперь используем [Power Analysis](https://en.wikipedia.org/wiki/Power_(statistics)) \([или вот](http://www.statskingdom.com/doc_test_power.html)\). По нему надо посчитать ![](https://wikimedia.org/api/rest_v1/media/math/render/svg/d7326aa9d5d7c4c8bcbb9c569949cd1e53203a88) Здесь **Ф** - это куммулятивное распределение нормальной функции, 1.64 - константа для 95% уверенности, после минуса - обратное стандартное отклонение (которое используется для расчета доверительного интервала), умноженное на дельту между средними значениями двух распределений. Это значение должно быть >95%, но если не вычитать из 1, то <5%.

Нулевая гипотеза состоит в том, что распределения не отличаются.

Она отвергается с 95% уверенность, если посчитанное значение <0.05.

````python
are_you_sure = {}

def calcStatPower(test_no):
    if test_no == 0:
        data2 = user_data[['parallel_lang', 'mean_mark']]
    else:
        data2 = user_data[['parallel_lang', 'mean_mark']][(user_data.test_id==test_no)]

    data = data2.groupby(['parallel_lang']).agg(['count', 'mean', 'std'])
    data['se'] = data[('mean_mark', 'std')] / np.sqrt(data[('mean_mark', 'count')])
    data['ci_min'] = data[('mean_mark', 'mean')] - 1.96 * data['se']
    data['ci_max'] = data[('mean_mark', 'mean')] + 1.96 * data['se']
    #data = data.drop([('uavg', 'std'), 'ci'], axis=1)
#     data.index = [i[1] for i in data.index]
    data.columns = ['count', 'mean', 'std', 'se', 'ci_min', 'ci_max']

    data["power"] = 0.
    for i in data.index:
        data.loc[i, "power"] = norm.cdf(1.645-abs(data['mean'][i]-
                                                  data['mean']['No Parallel Text'])/(data['se'][i]))

    for d, i in zip(data["power"], data["power"].index):
        are_you_sure[(i, test_no)] = d>0.05

    data_show = data.style.format({'power':'{:5.3f}'})
    display(data_show)
````
